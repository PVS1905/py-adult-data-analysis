# Task 0
Read the dataset from csv file & perform data cleaning - remove all rows, which contains `?` in some columns.
Also check for data correctness (salary & salary $K).

```python
people = pd.read_csv("/Users/viktor/advanced/py-adult-data-analysis/data/adult.csv", index_col=[0])
people.replace("?", np.nan, inplace=True)
people.dropna(inplace=True)

people.value_counts()
people.info()
```

# Task 4
Get the mean and standard deviation of the age for people who earn > 50K per year. After this, get it for those who earn <= 50K.

# Task 7
Compare the married and non-married men salaries. Who earns more? (>50K or <=50K)
Married men are those, whom `marital-status` starts with "Married". Others are not.

```python
married_status = [
    "Married-civ-spouse",
    "Married-AF-spouse",
    "Married-spouse-absent"
]

people_married = people[people["marital-status"].isin(married_status)]
people_not_married = people[~people["marital-status"].isin(married_status)]
people_married.groupby("salary")["age"].describe()
people_not_married.groupby("salary")["age"].describe()
```
